# Loading and exploring data with Pandas

**Pandas** is a library we've seen before for handling tabular data. Its alot more robust for data that is not just numbers than **NumPy**. Its main data format is the **DataFrame**, which is like a 2D array, but with column names and indexes for the rows. 

First we load in the data. This particular dataset is in the **tab-separated variable (.tsv)** format.

``
df = pd.read_csv("lbs.tsv", delimiter="\t")
``

Pandas shows us the **head** (first rows) and the **tail** (last rows), as well as the **shape**. So we know there are 16 columns (separate pieces of info about each entry), and 626 entries in total.

We can see its a mix of categorical data (``BuildingType``, ``County`` etc...), unique IDs, names, dates, monetary values and urls (a real mixed bag!). 

The categories are mainly **nominal**, in that they have no instrinsic order. Even the ones which are numbers (like ``PersonID``) are **nominal**, in that the numbers don't represent a ranking (one isn't best), and we would never want to do any maths with them. 

The dates and money however, are **continuous**, in that they are numbers that could take any value, and that we could do arithmetic (e.g. subtract one from the other to find a time).

```python
import pandas as pd
```

```python
df = pd.read_csv("lbs.tsv", delimiter="\t")
```

```python
df
```

### Data Types

We've considered what data is there, but some is numbers, some is text, some are dates. What does **Pandas** think each one is? This is important because it will determine what we can do with the data each column, how it will be sorted, and filtered etc...

We can use 

``df.dtypes``

to see each columns data type. We see that they are a mixed of ``object``s, ``int64`` and ``float64``.

Whilst we have seen ints and floats before, object is a new one!``object`` is the Pandas type for strings or mixed values. Do all of the columns that have ``object`` data type represent strings? 

Not all, ``Cost`` is an numerical value, but is shown as an ``object``. 

Why does having things like dates and money as numbers make our lives easier? This means we can do things like compare them (e.g. which one is earlier?), which is really useful for sorting, or do maths with them.

We'll investigate how to get numbers for ``Cost``later.

```python
df.dtypes
```

### Selecting Columns 

We can select a column using its name 

``df["ColumnName"]``

Or we can select a bunch of columns by passing an array 

``df[["ColumnName1","ColumnName2"]]``

This returns a smaller **Series** object with the results but we can also use 

``result.values``

To get an array, if, for example, we wanted to use some **NumPy** methods on it

After selecting a column, we can use describe to get just the stats for that data. 

We see a big divide in the ``Sex`` category, although this perhaps tells us more about property ownership historically in Britain in general.

### Counting Columns 

Beyond looking at the overall stats from `describe()`, we can use `value_counts()` to specifically count the number of times something appears in a column. This is similar to what we would see with a histogram (categories counted into bins).

``df["Category"].value_counts()``

This is useful, we can see the most commonly occuring family name (**Hibbert** and **Gordon**), and the most commonly occurring county (**Surrey**). We can even use two columns to find the most commonly occurring combinations. For example, we could try titles and counties. Its **Reverends** in **Surrey**.


``df[["Category","SubCategoryName"]].value_counts()``


```python
df["County"].value_counts()
```

```python
df["Title"].value_counts()
```

```python
#Instead we pass an array to get combinations of columns
df[["Title","County"]].value_counts()
```

### Sorting

Slavery was legally abolished in the UK in 1833. Given this, we could want to find out when the last property was built. 

We can sort the ``DataFrame`` by a column, then trim off the top 10. We see that actually there are two places in the dataset built in the 21 Century! Both are public memorials.

There is a then a drop off back to 1888 but still lots built after 1833, showing the continued impact long after abolition.

```python
df.sort_values(by='BuiltDate', ascending=False)[:10]
```

### Filtering

As well as picking whole columns, we can also pick columns that fit certain parameters, using **filtering**. To do this we pick columns that equal a certain value. 

Given `Rev.` was a popular title, lets see how many were owned by the **Church of England**. The **National Trust** is also a big owner of historic properties and is currently quite publically dealing with its own attempts to address colonial issues. 

Interestingly the Church specifically owns 6 properties, whereas its 25 for the National Trust.

```python
church_owned = df[df["Ownership"]=="Church of England"]
church_owned
```

```python
national_trust = df[df["Ownership"]=="National Trust"]
national_trust
```

```python
len(church_owned), len(national_trust)
```

### Missing Values

**NaN** in Pandas represents a missing value, or something that it is unable to format, and are excluded by **count()**

This means we can divide the ``count()`` by the ``len`` and anything that is not 1.00 will have missing values!

Given the way this dataset is collected, it is understandable that most columns will have missing values. 

We can use **isna()** to filter all the ones out to see if theres a reason why these are missing!

For example, ~2% of items dont have a building name. We can filter them out first

``df[df["BuildingName"].isna()]``

We can look at just the ``BuildingDescr`` column for those and see if that has information about the name in

``df[df["BuildingName"].isna()]["BuildingDescr"].values``

```python
#Divide without NaN by total length revealing which columns have missing values
df.count() / len(df)
```

```python
#Look at rows for which BuildingName(s) contains NaN
df[df["BuildingName"].isna()]["BuildingDescr"].values
```
